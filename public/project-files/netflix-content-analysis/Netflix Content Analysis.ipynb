# EDA (Exploratory Data Analysis)

```python
# Import visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Set theme
sns.set_theme(style="whitegrid")

# Figure size
plt.rcParams["figure.figsize"] = (12,6)

# Font sizes
plt.rcParams["axes.titlesize"] = 16
plt.rcParams["axes.labelsize"] = 12
plt.rcParams["xtick.labelsize"] = 10
plt.rcParams["ytick.labelsize"] = 10
```

```python
# EDA 1 — Movies vs TV Shows
# Business Question
# What percentage of Netflix's catalog consists of Movies and TV Shows?

# Count Movies and TV Shows
content_type = netflix['type'].value_counts()
print(content_type)
```

```python
# Visualization
plt.figure(figsize=(6,6))

content_type.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Netflix Content Type Distribution")
plt.ylabel("")
plt.show()
```

```python
# Explanation
# value_counts() counts each content type.
# kind='pie' creates a pie chart.
# autopct displays percentages.
# startangle=90 rotates the chart for better readability.
# ylabel("") removes the default y-axis label.
```

Business Insight

Movies account for approximately 70% of Netflix's catalog, while TV Shows make up around 30%. This indicates that Netflix's library is heavily focused on movies, although TV Shows remain a significant part of its streaming offering

# EDA 2 – Netflix Content Growth Over Time

Business Question

How has Netflix's content library grown over the years?

This analysis helps us understand:

When Netflix expanded rapidly.
Whether content additions are increasing or decreasing.
Trends in Netflix's content acquisition strategy.

```python
# Count content added each year
content_growth = (
    netflix['year_added']
    .value_counts()
    .sort_index()
)

print(content_growth)

# Explanation
# value_counts() counts the number of titles added each year.
# sort_index() sorts the years in ascending order.
```

```python
# Create the Line Chart
plt.figure(figsize=(12,6))

plt.plot(
    content_growth.index,
    content_growth.values,
    marker='o',
    linewidth=2
)

plt.title("Netflix Content Added by Year", fontsize=16)
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.xticks(content_growth.index.astype(int), rotation=45)

plt.grid(True)

plt.show()
```

Interpretation
Netflix experienced slow growth until 2015. Between 2016 and 2019, the number of titles added increased dramatically, reaching its highest point in 2019. After 2019, the number of additions declined slightly, suggesting a slowdown in content expansion.

Business Insight
Netflix invested heavily in expanding its content library from 2016 to 2019, likely to support rapid global subscriber growth. The slight decline after 2019 may reflect strategic changes in content production or external factors affecting releases.

# EDA 3 – Top 10 Countries
Business Question
Which countries contribute the most content to Netflix?

```python
# Split Multiple Countries
country_counts = (
    netflix['country']
    .str.split(', ', expand=True)
    .stack()
    .value_counts()
)

# Remove "Unknown"
country_counts = country_counts.drop('Unknown', errors='ignore')

# Select Top 10
top_countries = country_counts.head(10)
print(top_countries)

# Create the Chart
plt.figure(figsize=(10,6))

top_countries.sort_values().plot(
    kind='barh'
)

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

Why Horizontal Bar Chart?
Country names are long.Horizontal bars make them much easier to read.

Interpretation
The United States contributes the largest amount of Netflix content, followed by India, the United Kingdom, Canada, and France. This shows that Netflix's library is dominated by productions from a few major entertainment markets.

Business Insight
Although the United States remains Netflix's largest content source, international markets such as India, South Korea, and Spain contribute significantly, highlighting Netflix's investment in global content.

# Top Genres
Business Question
Which genres are most common on Netflix?

```python
# Split Genres
genre_counts = (
    netflix['listed_in']
    .str.split(', ', expand=True)
    .stack()
    .value_counts()
)

# Top 10 Genres
top_genres = genre_counts.head(10)
print(top_genres)

# Visualization
plt.figure(figsize=(10,6))

plt.hlines(
    y=top_genres.index,
    xmin=0,
    xmax=top_genres.values,
    color='lightgray'
)

plt.plot(
    top_genres.values,
    top_genres.index,
    "o",
    markersize=10
)

plt.title("Top Netflix Genres")

plt.xlabel("Number of Titles")

plt.show()
```

Interpretation
International Movies, Dramas, and Comedies are the most common genres available on Netflix, reflecting the platform's broad appeal to a global audience.

Business Insight
Netflix has built a diverse catalog with a strong emphasis on international productions and dramatic content, allowing it to serve audiences across multiple regions and preferences.

# EDA 5 – Content Ratings Distribution
Business Question.
What are the most common content ratings on Netflix?

```python
# Count each rating
rating_counts = (
    netflix["rating"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(13,6))

ax = sns.barplot(
    x=rating_counts.values,
    y=rating_counts.index,
    color="steelblue"
)

# Add labels
for i, value in enumerate(rating_counts.values):
    ax.text(
        value + 20,
        i,
        f"{value:,}",
        va="center",
        fontsize=10
    )

plt.title("Distribution of Netflix Content Ratings", weight="bold")
plt.xlabel("Number of Titles")
plt.ylabel("Content Rating")

plt.tight_layout()
plt.show()
```

Business Insight
TV-MA dominates Netflix's catalog, indicating that the platform primarily targets mature audiences. TV-14 is the second most common rating, suggesting a strong focus on teenage and young adult viewers.

# Advanced EDA 13 – Top 10 Countries by Content Type
Business Question: 
Which countries contribute the most Movies and TV Shows?

```python
# use the first listed country for this analysis to keep the visualization straightforward.
netflix['main_country'] = netflix['country'].str.split(',').str[0].str.strip()

country_type = (
    netflix.groupby(['main_country', 'type'])
    .size()
    .reset_index(name='count')
)

top_countries = (
    netflix['main_country']
    .value_counts()
    .head(10)
    .index
)

country_type = country_type[
    country_type['main_country'].isin(top_countries)
]

# Visualization
plt.figure(figsize=(14,7))

sns.barplot(
    data=country_type,
    x='main_country',
    y='count',
    hue='type'
)

plt.title("Top 10 Countries by Content Type", weight="bold")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Business Insigh: 
The United States contributes the largest number of both Movies and TV Shows, while other leading countries show different content production patterns.

# Advanced EDA 14 – Top Genres by Content Type
Business Question: 
Which genres are most common for Movies and TV Shows?

```python
genre_type = netflix.copy()

genre_type['genre'] = genre_type['listed_in'].str.split(',').str[0]

genre_summary = (
    genre_type.groupby(['genre', 'type'])
    .size()
    .reset_index(name='count')
)

top_genres = (
    genre_summary.groupby('genre')['count']
    .sum()
    .nlargest(10)
    .index
)

genre_summary = genre_summary[
    genre_summary['genre'].isin(top_genres)
]

# Visualization
plt.figure(figsize=(14,7))

sns.barplot(
    data=genre_summary,
    x='genre',
    y='count',
    hue='type'
)

plt.title("Top Genres by Content Type", weight="bold")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Business Insight: 
Movies dominate most genres, but TV Shows have a stronger presence in categories such as Kids' TV and International TV Shows.
